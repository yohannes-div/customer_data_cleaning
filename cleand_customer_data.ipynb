```
import pandas as pd
data = {"Name": ["john", "mary", "david", "john", "sara", None], "Email": ["JOHN@gmail.com", "mary@gmail.com", None, "JOHN@gmail.com", "sara@gmail.com", "david@gmail.com"], "Age": [25, 30, None, 25, 28, 35], "city": ["Addis Abeba", "Jimma", "Hawassa", "Addis Abeba", "jimma", "shashamane"]}
df = pd.DataFrame(data)
df
```

```
df["Name"] = df["Name"].str.strip().str.title()
display(df)
```

```
df["Email"]
df["Email"] = df["Email"].str.strip().str.lower()
display(df)
```

```
df["city"] = df["city"].str.strip().str.title()

display(df)
```

```
df = df.drop_duplicates()
display(df)
```

```
print(df.isnull().sum())
```

```
df = df.copy()
df.loc[:, "Name"] = df["Name"].fillna("unknown")
df.loc[:, "Email"] = df["Email"].fillna("not provided")
df.loc[:, "Age"] = df["Age"].fillna(df["Age"].median())
display(df)
```

```
df["Age"] = df["Age"].fillna(df["Age"].median())
display(df)
```

```
print(df.isnull().sum())
```

```
print("dataset shape:", df.shape)
print("duplicate rows:", df.duplicated().sum())
print("missing value:")
print(df.isnull().sum())
display(df)
```

```
df.to_csv("cleaned_customer_data.csv", index=False)
print("Dataset saved successfuly!")
```
